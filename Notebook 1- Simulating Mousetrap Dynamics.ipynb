## 1.1) Mathematical formalism
The most important formulas are the following and the explanation of what they represents will be subsequently given:
1. Branching Ratio - The expected number of new traps triggered per fired trap.   

$$ 
R_0 = p\,\rho\,\pi r_0^2 = p\,\rho\,A
$$

p = trigger probability per contact

$$ 
\rho 
$$ =Trap density

A = Area a fired ball can reach

2. Critical density (the threshold) - Below this density the cascades die out and above this density sustained chain reactions become possible. Derived by setting the branching ratio to one and obtaining the density.
$$
\rho_c = \frac{1}{p\,\pi r_0^2}
$$

3. The reaction-diffusion PDE (the main equation) - This models the evolution of a quantity that spreads through space and changes locally due to interactions or reactions. The equation for a single scalar field is the following:

$$
\frac{\partial u}{\partial t}
= D \nabla^2 u + R(u),
$$

However, in pattern formation, the system of coupled differential equations that describe the system is the following:

$$
\frac{\partial u}{\partial t}
= D_u \nabla^2 u + f(u,v)
$$

$$
\frac{\partial v}{\partial t}
= D_v \nabla^2 v + g(u,v)
$$

The two competing effects in this system are the following:
1. Diffusion - Spatial spreading from high to low concentration
2. Reaction - Local production, consumption or interaction between species

Here, there are two interacting species. A species is just something that is beaing measured,in nuclear diffusion it could be different type of nucleoid and in population synamics it could be different species of animals.

The meanings of the parameters are the following (keep in mind there are two species):
- u and v = Concentrations of each species
- D_u and D_v = Diffusion concentrations of each species, arbitrary constants
- x = Spatial coordinates
- f(u,v) and g(u,v) = Local reaction rate for u and v. The local reaction rate is how the substane is created / destroyed locally.

The $$ D_v \nabla^2 v $$ refers to the movement and spreading of v and  $$ g(u,v) $$ refers to the creation / destruction of v.
